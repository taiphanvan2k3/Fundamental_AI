```python
def byte_to_image(data, image_size=(64, 64)):
    """
    Chuyển đổi mảng NumPy chứa dữ liệu gói tin thành hình ảnh bằng cách coi các giá trị là byte.

    Args:
        data (np.ndarray): Mảng NumPy chứa dữ liệu gói tin.
        image_size (tuple): Kích thước mong muốn của hình ảnh (width, height).

    Returns:
        PIL.Image.Image: Hình ảnh được tạo ra.
    """

    # 1. Chuẩn hóa (tùy chọn)
    data = np.clip(data, 0, 255)  # Đảm bảo giá trị nằm trong [0, 255]
    data = data.astype(np.uint8)  # Chuyển đổi thành kiểu dữ liệu byte

    # 2. Reshape thành mảng 2D
    image_data = data.flatten()[:image_size[0] * image_size[1]] # Chỉ lấy số lượng cần thiết

    #Nếu kích thước mảng không đủ, thêm các giá trị 0
    if image_data.shape[0] < image_size[0] * image_size[1]:
        padding_size = image_size[0] * image_size[1] - image_data.shape[0]
        image_data = np.concatenate([image_data, np.zeros(padding_size)])

    image_data = image_data.reshape(image_size)

    # 3. Tạo hình ảnh PIL
    img = image_data.astype(np.uint8)  # Đảm bảo kiểu dữ liệu phù hợp
    return img
```

```python
image = byte_to_image(X1, image_size=(128, 128))
cv2.imshow("Byte Image", image) # Hiển thị ảnh dùng cv2
cv2.waitKey(0) # Đợi đến khi có phím bấm
cv2.destroyAllWindows() # Đóng tất cả các cửa sổ
```

```python
import matplotlib.pyplot as plt

def byte_to_image_per_packet_plt(data, image_size=(1, 10), num_to_show=10):
    """
    Chuyển đổi và hiển thị một số hình ảnh gói tin đầu tiên bằng matplotlib.

    Args:
        data (np.ndarray): Mảng NumPy, mỗi hàng đại diện cho một gói tin.
        image_size (tuple): Kích thước mong muốn của mỗi hình ảnh (height, width).
        num_to_show (int): Số lượng hình ảnh muốn hiển thị.
    """
    num_packets = min(data.shape[0], num_to_show)

    for i in range(num_packets):
        packet_data = data[i, :]
        packet_data = np.clip(packet_data, 0, 255)
        packet_data = packet_data.astype(np.uint8)

        if packet_data.shape[0] < image_size[0] * image_size[1]:
            padding_size = image_size[0] * image_size[1] - packet_data.shape[0]
            packet_data = np.concatenate([packet_data, np.zeros(padding_size)])

        packet_data = packet_data[:image_size[0] * image_size[1]].reshape(image_size)

        plt.imshow(packet_data, cmap='gray')
        plt.title(f"Packet {i}")
        plt.axis('off')
        plt.show()

# Mỗi record có 10 columns, ta sẽ chuyển đổi thành hình ảnh 1x10
image_size = (1, 10)
byte_to_image_per_packet_plt(X1, image_size=image_size, num_to_show=5)
```

```python
def byte_to_image(X, shape=(2, 5), scale_range=(0, 255), to_rgb=False):
    """
    Chuyển dữ liệu tabular thành ảnh theo phương pháp byte-to-image.
    
    Args:
        X (np.ndarray): Dữ liệu đầu vào, dạng (samples, features)
        shape (tuple): Kích thước ảnh (height, width) sau reshape
        scale_range (tuple): Khoảng giá trị (min, max) để scale về 0-255
        to_rgb (bool): Nếu True thì convert sang ảnh RGB

    Returns:
        np.ndarray: Ảnh đầu ra dạng (samples, height, width, channels)
    """
    from sklearn.preprocessing import MinMaxScaler

    scaler = MinMaxScaler(feature_range=scale_range)
    X_scaled = scaler.fit_transform(X).astype(np.uint8)
    
    h, w = shape
    assert X.shape[1] == h * w, "Số feature không khớp với kích thước ảnh mong muốn"

    X_img = X_scaled.reshape(-1, h, w, 1)

    if to_rgb:
        X_img = np.repeat(X_img, 3, axis=-1)  # chuyển từ 1 kênh sang 3 kênh

    return X_img
```

```python
X_img = byte_to_image(X1, shape=(2, 5), to_rgb=False)
print(X_img.shape)  # (225745, 2, 5, 1)
```
