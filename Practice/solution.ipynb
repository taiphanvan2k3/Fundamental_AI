```python
def transform_to_image(data, img_size=32):
    num_features = data.shape[1]
    while img_size * img_size < num_features:
        img_size += 1
    images = []
    for index in data.index:
        sample = data.loc[index].values  # Get values as a NumPy array
        # Pad if necessary
        if img_size * img_size > num_features:
            padded = np.pad(sample, (0, img_size * img_size - num_features), "constant")
        else:
            padded = sample[: img_size * img_size]

        # Reshape into a 2D image
        img = padded.reshape((img_size, img_size))
        images.append(img)
    return np.array(images)


img_size = 9
X_train_img = transform_to_image(X_train, img_size)
X_test_img = transform_to_image(X_test, img_size)

# Add a color channel to match the CNN input
X_train_img = X_train_img[..., np.newaxis]
X_test_img = X_test_img[..., np.newaxis]
```

### 1. Byte to image


```python
from sklearn.preprocessing import MinMaxScaler
```

```python
def byte_to_image(data, shape=(8, 8), to_rgb=False):
    """
    Chuyển đổi dữ liệu số học thành ảnh (giá trị uint8 từ 0-255).

    Args:
        data (np.ndarray): Dữ liệu đầu vào dạng 2D (samples, features).
        shape (tuple): Kích thước ảnh đầu ra (height, width).
                       Tổng height*width phải >= số lượng feature.
        to_rgb (bool): Nếu True, tạo ảnh RGB (3 channels). Mặc định: False (grayscale).

    Returns:
        np.ndarray: Ảnh đầu ra dạng (num_samples, H, W, C).
    """
    num_samples, num_features = data.shape
    img_height, img_width = shape
    num_pixels = img_height * img_width

    if num_pixels < num_features:
        raise ValueError("shape quá nhỏ để chứa toàn bộ feature!")

    # 1. Scale về [0, 255]
    scaler = MinMaxScaler(feature_range=(0, 255))
    data_scaled = scaler.fit_transform(data).astype(np.uint8)

    # 2. Padding nếu thiếu pixel để reshape thành ảnh
    if num_pixels > num_features:
        pad_width = num_pixels - num_features
        data_scaled = np.pad(data_scaled, ((0, 0), (0, pad_width)), mode="constant")

    # 3. Reshape thành ảnh
    images = data_scaled.reshape((num_samples, img_height, img_width))

    # 4. Thêm channel dimension
    if to_rgb:
        images = np.stack([images] * 3, axis=-1)  # RGB ảnh 3 channel
    else:
        images = images[..., np.newaxis]  # Grayscale ảnh 1 channel

    return images
```

```python
import matplotlib.pyplot as plt


def show_images(images, num=5, cmap="gray"):
    """
    Hiển thị một vài ảnh đầu tiên từ mảng ảnh 4D.

    Args:
        images (np.ndarray): Mảng ảnh có shape (samples, H, W, C)
        num (int): Số lượng ảnh muốn hiển thị
        cmap (str): Màu sắc hiển thị ('gray' cho grayscale, None cho RGB)
    """
    plt.figure(figsize=(num * 2, 2))
    for i in range(num):
        plt.subplot(1, num, i + 1)
        img = images[i]
        if img.shape[-1] == 1:
            img = img.squeeze()  # bỏ dimension channel nếu grayscale
        plt.imshow(img, cmap=cmap)
        plt.axis("off")
    plt.tight_layout()
    plt.show()


show_images(X_img, num=5, cmap="gray")
```

### 2. Mapping


```python
def map_features_to_image(X, img_size=4, mapping_positions=None):
    num_samples, num_features = X.shape
    image_data = np.zeros((num_samples, img_size, img_size), dtype=np.float32)

    scaler = MinMaxScaler(feature_range=(0, 255))
    data_scaled = scaler.fit_transform(X).astype(np.uint8)

    # Nếu không cung cấp vị trí mapping, mặc định là từ trái sang phải
    if mapping_positions is None:
        flat_positions = [(i // img_size, i % img_size) for i in range(num_features)]
    else:
        flat_positions = mapping_positions

    for idx, (row, col) in enumerate(flat_positions):
        image_data[:, row, col] = data_scaled[:, idx]

    return image_data[..., np.newaxis]  # Thêm kênh màu


# Demo
# X: (5600, 78)
X_mapped = map_features_to_image(X, img_size=9)
print(X_mapped.shape)
```

### Training


```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, Input
from tensorflow.keras.optimizers import Adam
```

```python
def build_simple_cnn(img_size):
    cnn_model = Sequential()
    cnn_model.add(Input(shape=(img_size, img_size, 1)))  # Input layer with grayscale image

    cnn_model.add(Conv2D(32, (3, 3), activation='relu', padding='same'))
    cnn_model.add(MaxPooling2D((2, 2)))

    cnn_model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    cnn_model.add(MaxPooling2D((2, 2)))

    cnn_model.add(Conv2D(128, (3, 3), activation='relu', padding='same'))
    cnn_model.add(MaxPooling2D((2, 2)))

    cnn_model.add(Flatten())
    cnn_model.add(Dense(128, activation='relu'))

    cnn_model.add(Dropout(0.5))
    cnn_model.add(Dense(4, activation='softmax'))  # 4 classes: BENIGN, DDoS, Brute Force, DoS

    return cnn_model
```

```python
cnn_model = build_simple_cnn(img_size=9)

cnn_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"],
)

cnn_model.summary()
```

### Dùng với ResNet50

```python
from tensorflow.keras.applications.resnet50 import preprocess_input
from tensorflow.image import resize

def prepare_resnet_input(images):
    # Convert grayscale to RGB
    images_rgb = tf.image.grayscale_to_rgb(tf.convert_to_tensor(images, dtype=tf.float32))
    # Resize to 224x224
    images_resized = resize(images_rgb, [112, 112])
    # Normalize theo chuẩn ResNet
    return preprocess_input(images_resized)
```

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout, Input

def build_resnet_model(num_classes=4):
    base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(112, 112, 3))

    # Freeze base
    for layer in base_model.layers:
        layer.trainable = False

    # Thêm tầng phân loại
    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation='relu')(x)
    x = Dropout(0.5)(x)
    predictions = Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=base_model.input, outputs=predictions)
    return model
```

```python
resnet_model = build_resnet_model()
resnet_model.compile(optimizer=Adam(1e-4), loss='categorical_crossentropy', metrics=['accuracy'])

resnet_model.summary()
```

```python
def preprocess_image(image, label):
    image = tf.image.grayscale_to_rgb(image)
    image = tf.image.resize(image, [112, 112])
    image = preprocess_input(image)
    return image, label
```

```python
train_ds = tf.data.Dataset.from_tensor_slices((X_train_img, y_train))
train_ds = train_ds.map(preprocess_image).batch(32).prefetch(tf.data.AUTOTUNE)

test_ds = tf.data.Dataset.from_tensor_slices((X_test_img, y_test))
test_ds = test_ds.map(preprocess_image).batch(32).prefetch(tf.data.AUTOTUNE)
```

```python
history = resnet_model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=10
)
```
